```python
import sympy
import math


# Переводит строку в список кодов символов через ord()
def text_to_code(text):
  code_text = []
  for letter in text:
    code_text.append(ord(letter))
  return code_text

# Превращает обратно список чисел в строку через chr()
def code_to_text(list_code):
  text = ''
  for number in list_code:
    text += chr(number)
  return text

# Генерирует два разных простых числа
def gen_prime(limit):
  p = sympy.randprime(0, limit)
  q = p
  while q == p:
    q = sympy.randprime(0, limit)
  return p, q

# Ищет наименьшее e, взаимно простое с φ(N), начиная с e=2
def gen_e(phi):
  e = 2
  while(e < phi):
    if math.gcd(e, phi) == 1:
      return e
    else:
      e += 1

# Создает открытый и закрытый ключи RSA
def gen_keys(lim_for_prime = 1000): # верхняя граница для поиска простых чисел, то есть p и q могут быть простыми числами меньше 1000
  p, q = gen_prime(limit = lim_for_prime)
  N = p*q # открытый модуль RSA, входящий в открытый и закрытый ключи
  phi = (p-1)*(q-1) # функция Эйлера, позволяющая подобрать пару e и d, которые взаимно отменяют друг дурга при возведении в степен по модулю N
  e = gen_e(phi)
  d = pow(e, -1, phi)
  return {'open': (e, N), 'close': (d, N)}

# Шифрует текст открытым ключом, возводя код символа в степень e по модулю N
def shifr_message(text, open_key):
  e = open_key[0]
  mod = open_key[1]
  shifr = []
  code = text_to_code(text)
  for elem in code:
    shifr.append(elem ** e % mod)
  return shifr

# Расшифровывает список чисел закрытым ключом, возводя каждоче число в степен d по модулю N
def deshifr_message(shifr, close_key):
  d = close_key[0]
  mod = close_key[1]
  code = []
  for elem in shifr:
    code.append(elem ** d % mod)
    text = code_to_text(code)
  return text

def main(): # начало программы
  keys = gen_keys(1000)
  phrase = 'Привет, как дела? То, что ты просил: 123421' # сообщение для шифровки

  encrypted = shifr_message(phrase, keys['open'])
  decrypted = deshifr_message(encrypted, keys['close'])


your_keys = gen_keys(lim_for_prime=300)
print(f"Открытый ключ Жени: {your_keys['open']}")
print(f"Закрытый ключ Жени: {your_keys['close']}")
print(f"Сообщение Жени: Настя, отправь номер карты, мой открытый ключ {your_keys['open']}")
answer = 'Номер - 4276390142335562'
print(f"Ответ Насти: {answer}")
shifr_answer = shifr_message(answer, your_keys['open'])
print(f"Зашифрованный ответ Насти: {shifr_answer}")
deshifr_answer = deshifr_message(shifr_answer, your_keys['close'])
print(f"Расшифрованный ответ Насти: {deshifr_answer}")
```
